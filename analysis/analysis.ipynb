# Flight analysis
The latest recorded run (set `RUN` for an older one). **Run all cells.** Notes on the data: at the end.

````python
RUN = None   # None = the latest run, or e.g. '2026-09-25_14-32-07'

import matplotlib.pyplot as plt
import runlog

run = runlog.load_run(RUN)
runlog.show_objective(run)
````

## Summary

````python
runlog.print_summary(run)
````

## Where did it fly?
Top view with the balloons (popped ones faded; red ones outlined, a penalty) and where the drone was every 5 s; the altitude against the balloons' height band; the distance to the nearest balloon of each colour (how close it came to a target, and to a red one).

````python
runlog.plot_overview(run)
plt.show()
````

## What the camera could see
For every balloon: when it was inside the camera's view and how big it looked (worked out from the ground truth, so balloons hidden behind others still count). **A good balloon in view for a long time and never chased is a missed chance**; a growing bar is the drone closing in.

````python
runlog.plot_visibility(run)
plt.show()
````

## Your commands
What reached the drone from your controller (after the failsafe, if it kicked in).

````python
runlog.plot_commands(run)
plt.show()
````

## The camera at key moments
What your controller was looking at: after arming, just before each pop, at the end.
`runlog.show_frames(run, [('my label', 7.5)])` shows any other time.

````python
runlog.show_frames(run)
plt.show()
````

## Diagnostics: attitude, flight controller vs truth
Your controller only sees the flight controller's estimates. Usually they are within a degree or two of the truth; a large gap means the estimate was off.

````python
runlog.plot_attitude(run)
plt.show()
````

## About this notebook

Looks at a recorded run from `analysis/runs/`. **Run all cells** (Kernel → Restart & Run All).
By default it shows the **latest** run; set `RUN` in the first code cell to look at an older one.

Every competition launch records itself automatically:
```
ros2 launch pluto_x_bringup competition.launch.py controller:=outerloop_controller/my_controller.py
```
The balloon layout comes from `analysis/seed.yaml`. It stays the same between runs, so you can compare
changes to your controller. Get a new layout with `python3 analysis/new_seed.py`.

Note: the trajectory is the simulator's **ground truth**, for analysis only. Your controller never sees it.
