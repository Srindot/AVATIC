# Evaluation report

Results of `evaluation/evaluate.py`: your controller on many random balloon layouts, like hidden test
cases. **Run all cells.** It shows the **latest** session; set `SESSION` to look at an older one.

````python
SESSION = None   # None = the latest session, or e.g. '2026-09-25_14-32-07'

import matplotlib.pyplot as plt
import evallog

session = evallog.load_session(SESSION)
evallog.show_objective(session)
````

## Points per run
Each bar is one random layout, split by colour: good balloons stack up from 0, red hits (−75 each) hang below it. The black dash is the run's score, the dashed line the mean.

````python
evallog.plot_scores(session)
plt.show()
````

## Score during the run
Every run's score against the time since arming (thin lines) and the mean (thick). A steep start means balloons are found fast; a flat end means the controller ran out of ideas, not out of balloons.

````python
evallog.plot_score_over_time(session)
plt.show()
````

## Every run

````python
evallog.table(session)
````

## The best and the worst run
Top view (ground truth, judges only). For any single run, all the plots of the analysis notebook work:
`run = evallog.load_run(session, 3)` then e.g. `evallog.runlog.plot_overview(run)`, `evallog.runlog.show_frames(run)`.

````python
evallog.plot_maps(session, ('best', 'worst'))
plt.show()
````

````python
runlog = evallog.runlog
worst = evallog.load_run(session, 'worst')
if worst is None:
    print('no recorded run in this session')
else:
    runlog.show_objective(worst)
    runlog.show_frames(worst)
    plt.show()
````
